```python
# =============================================================================
#
#   MULTIAGENT CRYOCOOLER CONTROL SYSTEM
#   =====================================
#   What this program does (plain English):
#
#   A cryocooler is a machine that cools things to very low temperatures
#   (like cooling a superconductor or a space sensor).
#
#   We simulate 5 agents (virtual robots/controllers):
#     - 1 LEADER  : does NOT control a real machine.
#                   It just creates the desired temperature path
#                   (called a "trajectory") that everyone else should follow.
#     - 4 FOLLOWERS : each controls its own cryocooler machine and tries
#                     to match the leader's temperature trajectory.
#
#   Each follower uses TWO controllers working together:
#     1. RISE  : a math-based robust controller that handles disturbances.
#     2. DNN   : a small neural network that learns the machine's behaviour
#                and helps the RISE controller compensate for it.
#
#   At the end, the program draws 4 figures, including a 3D plot showing
#   how well each follower tracked the leader over time.
#
#   HOW TO RUN:
#     1. Open Command Prompt
#     2. Type:  pip install torch numpy matplotlib
#     3. Type:  python multiagent_cryocooler_rise_dnn_explained.py
#
# =============================================================================


# =============================================================================
# STEP 1 — IMPORT LIBRARIES
# =============================================================================
# In Python, a "library" (also called a "module" or "package") is a collection
# of pre-written code that gives us extra tools.  We import them so we can use
# those tools without writing everything from scratch.

import numpy as np
# numpy  →  the standard library for maths and arrays in Python.
# "as np" means we can write np.sin() instead of numpy.sin() — just shorter.

import matplotlib.pyplot as plt
# matplotlib is the main plotting library in Python.
# pyplot is the sub-module that gives us plt.plot(), plt.show(), etc.

from mpl_toolkits.mplot3d import Axes3D
# This import "registers" the 3D projection so matplotlib can draw 3D axes.
# We never call Axes3D directly, but importing it is required.

from matplotlib import cm
# cm  =  "colour maps".  Used to pick colour schemes for 3D surfaces
# (e.g.  cm.coolwarm  gives a blue-white-red gradient).

from matplotlib.colors import Normalize
# Normalize maps a range of numbers to [0, 1] so colour maps work correctly.

import torch
# PyTorch is the deep-learning library.  We use it to build and train
# the neural network (DNN) inside each follower's controller.

import torch.nn as nn
# nn  =  "neural network".  Contains building blocks like Linear layers,
# activation functions (Tanh), and loss functions (MSELoss).

import torch.optim as optim
# optim  =  "optimiser".  Contains algorithms like Adam that adjust the
# neural network's weights during training to reduce the error.

import warnings
warnings.filterwarnings("ignore")
# This silences minor warning messages so the output stays clean.


# =============================================================================
# STEP 2 — SET RANDOM SEEDS (for reproducibility)
# =============================================================================
# Neural networks start with random weights.  By fixing the "seed" we make
# sure every run gives the same result, which is important for research.

torch.manual_seed(0)   # fixes randomness in PyTorch
np.random.seed(42)     # fixes randomness in NumPy


# =============================================================================
# STEP 3 — SIMULATION TIME SETTINGS
# =============================================================================
# We will simulate the system step-by-step, like a video game updating every
# fraction of a second.  Each update is one "time step".

DT    = 0.005    # size of each time step = 0.005 seconds (5 milliseconds)
T_SIM = 25.0     # total simulation time  = 25 seconds

# np.arange creates an array of evenly spaced numbers.
# Here: [0.000, 0.005, 0.010, ..., 24.995]
t_vec = np.arange(0.0, T_SIM, DT)

N = len(t_vec)   # total number of time steps  (25 / 0.005 = 5000 steps)

N_AGENTS = 4     # we have 4 follower agents (plus 1 leader defined later)


# =============================================================================
# STEP 4 — DEFINE THE PLANT (the physical cryocooler machine)
# =============================================================================
# The "plant" is the real physical system being controlled — here, a cryocooler.
#
# We model it as a second-order nonlinear differential equation
# (called a Duffing-type oscillator):
#
#   x''(t)  =  -delta*x'  -  alpha*x  -  beta*x^3  +  u  +  d(t)
#
# Where:
#   x       = temperature deviation from setpoint (Kelvin)
#   x'      = rate of temperature change (K/s)   [first derivative]
#   x''     = acceleration of temperature change  [second derivative]
#   delta   = damping coefficient  (thermal resistance)
#   alpha   = linear stiffness     (thermal capacitance effect)
#   beta    = cubic nonlinearity   (nonlinear thermal behaviour)
#   u       = control input        (compressor power, in Watts)
#   d(t)    = external disturbance (unwanted heat load)
#
# IMPORTANT: The controller does NOT know delta, alpha, beta — it must work
# without knowing the exact model.  That's why we need RISE + DNN.

# ---- Nominal (average) parameter values ----
delta_nom = 0.5   # nominal damping
alpha_nom = 4.0   # nominal linear stiffness
beta_nom  = 0.3   # nominal cubic nonlinearity

# ---- Per-follower parameter variations (+- 15%) ----
# Each real cryocooler is slightly different due to manufacturing tolerances.
# We model this by giving each follower slightly different delta/alpha/beta.

_v = 0.15   # 15% maximum variation

# These arrays define HOW MUCH each follower deviates from nominal.
# Follower 1 uses index 0, Follower 2 uses index 1, etc.
_dv = np.array([ 0.10, -0.12,  0.15, -0.08])   # delta variations
_av = np.array([-0.08,  0.10, -0.15,  0.12])   # alpha variations
_bv = np.array([ 0.12, -0.10,  0.08, -0.15])   # beta  variations

# Compute the actual parameter for each follower.
# Example for Follower 1:  delta = 0.5 * (1 + 0.15 * 0.10) = 0.5075
delta_arr = delta_nom * (1.0 + _v * _dv)
alpha_arr = alpha_nom * (1.0 + _v * _av)
beta_arr  = beta_nom  * (1.0 + _v * _bv)


def plant_f(x, xdot, i):
    """
    Computes the true nonlinear dynamics  f(x, x')  for follower i.

    This is the part of the physics that the controller does NOT know.
    The DNN will try to learn and approximate this function.

    Parameters
    ----------
    x    : float  — current temperature state
    xdot : float  — current temperature rate
    i    : int    — follower index (0 to 3)

    Returns
    -------
    float  — the value of f(x, x')
    """
    return -delta_arr[i] * xdot - alpha_arr[i] * x - beta_arr[i] * x ** 3
    # x**3 means x cubed  (x to the power of 3)


def disturbance(t, i):
    """
    Computes the external disturbance d(t) for follower i at time t.

    This represents unwanted heat leaking into the cryocooler from outside,
    which varies with time and is different for each machine.

    The controller must reject this disturbance — the RISE controller is
    specifically designed to handle bounded disturbances like this.
    """
    return (0.08 * np.sin(3.0 * t + i * np.pi / 3)
            + 0.05 * np.cos(7.0 * t + i * np.pi / 5))
    # np.pi is the mathematical constant pi (3.14159...)


# =============================================================================
# STEP 5 — DEFINE THE LEADER TRAJECTORY
# =============================================================================
# The leader agent generates the "desired" temperature path that all followers
# must track.  It is a smooth, multi-frequency sinusoidal signal.
#
# We need three things at every time step t:
#   xd(t)   = desired position  (temperature)
#   xd'(t)  = desired velocity  (rate of temperature change)
#   xd''(t) = desired acceleration
#
# These are needed by the RISE controller to compute tracking errors.

def leader_traj(t):
    """
    Returns the leader's desired trajectory at time t.

    The trajectory is:
        xd(t) = 1.5*sin(0.4*t) + 0.4*sin(1.8*t)

    This is a sum of two sine waves at different frequencies.
    Using two frequencies (0.4 and 1.8 rad/s) makes the trajectory
    more interesting and tests the controller's ability to track
    non-trivial signals.

    The velocity xd'(t) and acceleration xd''(t) are the first and
    second derivatives, computed using calculus rules:
        d/dt [A*sin(w*t)] = A*w*cos(w*t)
        d/dt [A*cos(w*t)] = -A*w*sin(w*t)
    """
    xd   =  1.5 * np.sin(0.4 * t) + 0.4 * np.sin(1.8 * t)
    xdd  =  1.5 * 0.4 * np.cos(0.4 * t) + 0.4 * 1.8 * np.cos(1.8 * t)
    xddd = -1.5 * 0.16 * np.sin(0.4 * t) - 0.4 * 3.24 * np.sin(1.8 * t)
    return xd, xdd, xddd   # return all three values at once (a "tuple")


# =============================================================================
# STEP 6 — BUILD THE DEEP NEURAL NETWORK (DNN)
# =============================================================================
# A neural network is a mathematical function made of many layers of numbers
# called "weights".  By adjusting the weights, the network can learn to
# approximate almost any function.
#
# Here, each follower has its own DNN that learns to approximate the unknown
# plant nonlinearity f(x, x').
#
# ARCHITECTURE:
#   Input  layer : 2 neurons   → receives [x, x'] (temperature + rate)
#   Hidden layer : 64 neurons  → with Tanh activation (squishes values to -1..1)
#   Hidden layer : 64 neurons  → with Tanh activation
#   Hidden layer : 32 neurons  → with Tanh activation
#   Output layer : 1 neuron    → outputs the estimated f_hat(x, x')
#
# Tanh is used instead of ReLU because the plant dynamics are smooth
# and Tanh produces smoother outputs, which is better for control.

class DynamicsNet(nn.Module):
    """
    A simple feedforward neural network that estimates f(x, x').

    In Python, 'class' defines a new type of object.
    'nn.Module' means this class is a PyTorch neural network.
    """

    def __init__(self, hidden=64):
        """
        __init__ is the constructor — it runs once when we create a DynamicsNet.
        'hidden' is how many neurons are in each hidden layer (default: 64).
        """
        super().__init__()   # always required when inheriting from nn.Module

        # nn.Sequential chains layers in order: input flows through each layer.
        self.net = nn.Sequential(
            nn.Linear(2,      hidden),  # fully-connected layer: 2 inputs -> 64
            nn.Tanh(),                  # activation function (nonlinearity)
            nn.Linear(hidden, hidden),  # 64 -> 64
            nn.Tanh(),
            nn.Linear(hidden, hidden // 2),  # 64 -> 32  (//)=integer division
            nn.Tanh(),
            nn.Linear(hidden // 2, 1),  # 32 -> 1 output
        )

    def forward(self, x):
        """
        forward() defines how data flows through the network.
        PyTorch calls this automatically when you "call" the network like a function.

        x      : a tensor (matrix) of shape [batch_size, 2]
        returns: a tensor of shape [batch_size]  (one output per sample)
        """
        return self.net(x).squeeze(-1)
        # .squeeze(-1) removes the last dimension of size 1
        # e.g. shape [1,1] becomes [1]


# Create one DNN for each follower agent.
# Python list comprehension: [expression  for variable  in range]
# This creates a list of 4 DynamicsNet objects.
dnns = [DynamicsNet() for _ in range(N_AGENTS)]

# Create one Adam optimiser for each DNN.
# Adam is an algorithm that updates the DNN weights to reduce prediction error.
# lr = "learning rate" — how big each weight update is (smaller = more stable)
optims = [optim.Adam(dnn.parameters(), lr=5e-4) for dnn in dnns]
# 5e-4 means 5 * 10^(-4) = 0.0005  (scientific notation)


# =============================================================================
# STEP 7 — BUILD THE RISE CONTROLLER
# =============================================================================
# RISE stands for: Robust Integral of the Sign of the Error
#
# It is a continuous, robust feedback controller designed for systems with
# unknown dynamics and bounded disturbances.
#
# HOW IT WORKS (step by step):
#
#   1. Compute tracking errors:
#        e1 = xd - x       (how far is temperature from desired?)
#        e2 = xd' - x'     (how far is temperature rate from desired?)
#
#   2. Compute auxiliary (filtered) error:
#        r = e2 + alpha * e1
#        (mixing position and velocity error into one signal)
#
#   3. Update the control law (discrete version, computed each time step):
#        u_RISE[k] = u_RISE[k-1]
#                  + (ks + 1) * (r[k] - r[k-1])   <- proportional to change in r
#                  + beta * sign(r[k]) * dt        <- integral of sign(r)
#
#      sign(r) = +1 if r > 0,  -1 if r < 0,  0 if r = 0
#      The integral of sign(r) is what makes RISE "robust" —
#      it continuously pushes the error toward zero.
#
#   4. Combine with DNN:
#        u_total = u_RISE - f_hat
#        (the DNN cancels out the estimated nonlinear dynamics,
#         so RISE only has to handle the residual error)
#
#   5. Saturate (clip) the output to physical actuator limits:
#        u_total = clip(u_total, -30, +30) Watts

class RISEController:
    """
    RISE controller for one follower agent.

    Parameters
    ----------
    ks    : proportional gain on the change in auxiliary error (default 6)
    alpha : weight mixing e1 into auxiliary signal r (default 3)
    beta  : gain on the integral-of-sign term (default 8)
    """

    def __init__(self, ks=6.0, alpha=3.0, beta=8.0):
        self.ks    = ks
        self.alpha = alpha
        self.beta  = beta

        # These store the controller's memory from the previous time step.
        # A controller that uses memory is called a "dynamic" controller.
        self._u_rise_prev = 0.0   # u_RISE at the previous step
        self._r_prev      = 0.0   # auxiliary error r at the previous step

    def step(self, e1, e2, f_hat, dt):
        """
        Computes the control output for one time step.

        Parameters
        ----------
        e1    : float — position error  (xd - x)
        e2    : float — velocity error  (xd' - x')
        f_hat : float — DNN estimate of f(x, x')
        dt    : float — time step size (seconds)

        Returns
        -------
        u_total : float — control signal sent to the plant (clipped to +-30 W)
        r       : float — auxiliary error signal (useful for diagnostics)
        """
        # Step A: compute auxiliary error
        r  = e2 + self.alpha * e1

        # Step B: compute change in r since last step
        dr = r - self._r_prev

        # Step C: RISE update rule (recursive integral form)
        u_rise = (self._u_rise_prev          # carry forward previous value
                  + (self.ks + 1) * dr       # proportional to rate of r
                  + self.beta * np.sign(r) * dt)   # integral of sign(r)

        # Step D: subtract DNN estimate to cancel known nonlinearity
        u_total = u_rise - f_hat

        # Step E: clip to actuator limits (cryocooler compressor power range)
        u_total = np.clip(u_total, -30.0, 30.0)
        # np.clip(value, min, max) forces a value to stay between min and max

        # Step F: save current values for next time step
        self._u_rise_prev = u_rise
        self._r_prev      = r

        return u_total, r   # return both (comma = Python tuple)


# Create one RISE controller per follower (all with the same gains)
controllers = [RISEController(ks=6, alpha=3, beta=8) for _ in range(N_AGENTS)]


# =============================================================================
# STEP 8 — ALLOCATE STORAGE ARRAYS
# =============================================================================
# Before running the simulation loop, we pre-allocate (reserve space for)
# arrays to store every value at every time step.
#
# np.zeros((rows, cols)) creates a matrix filled with zeros.
# We fill them in during the simulation loop below.

# State of each follower over time: position and velocity
# Shape: (4 agents, 5000 time steps, 2 states)
X = np.zeros((N_AGENTS, N, 2))
# X[i, k, 0]  = position  of follower i at time step k
# X[i, k, 1]  = velocity  of follower i at time step k

E1     = np.zeros((N_AGENTS, N))   # position tracking error per agent per step
E2     = np.zeros((N_AGENTS, N))   # velocity tracking error per agent per step
U      = np.zeros((N_AGENTS, N))   # control input applied
F_HAT  = np.zeros((N_AGENTS, N))   # DNN's estimate of f(x, x')
F_TRUE = np.zeros((N_AGENTS, N))   # true f(x, x') (for comparison in plots)


# =============================================================================
# STEP 9 — SET INITIAL CONDITIONS
# =============================================================================
# At time k=0 each follower starts at a slightly different temperature,
# offset from where the leader starts.  This tests whether the controller
# can drive the error to zero from a non-zero starting point.

IC_POS = np.array([0.40, -0.50,  0.60, -0.30])  # initial positions (K)
IC_VEL = np.array([0.00,  0.10, -0.10,  0.05])  # initial velocities (K/s)

for i in range(N_AGENTS):
    # range(N_AGENTS) produces [0, 1, 2, 3]
    # X[i, 0] refers to follower i at time step 0
    X[i, 0] = [IC_POS[i], IC_VEL[i]]   # set both states at once


# =============================================================================
# STEP 10 — HYPERPARAMETERS FOR ONLINE DNN TRAINING
# =============================================================================
DNN_UPDATE_EVERY = 5
# Update the DNN weights every 5 time steps (not every step, to save compute).

DNN_WARMUP_STEPS = 30
# Skip the first 30 steps before training the DNN.
# During the very first steps the states are still very transient
# (large rapid changes), which would give the DNN noisy training data.


# =============================================================================
# STEP 11 — MAIN SIMULATION LOOP
# =============================================================================
# This is the heart of the program.  We loop over every time step k,
# and for each follower i we:
#   (a) compute errors against the leader
#   (b) ask the DNN for f_hat
#   (c) compute the RISE+DNN control input
#   (d) advance the plant by one time step (Euler integration)
#   (e) update the DNN weights using what we just observed

print(f"\n{'='*60}")
print(f"  Multiagent Cryocooler Simulation")
print(f"  Leader + {N_AGENTS} Followers | RISE + DNN | T={T_SIM}s dt={DT}s")
print(f"{'='*60}")
print("  Running simulation ...", end="", flush=True)
# end="" means don't go to a new line yet
# flush=True forces the text to appear immediately

for k in range(N - 1):
    # range(N-1) goes from 0 to N-2.
    # We stop one step early because we write to k+1 inside the loop,
    # and we don't want to go past the end of the array.

    t_k = t_vec[k]   # current time (seconds)

    # Ask the leader for the desired trajectory at this moment
    xd, xdd, _ = leader_traj(t_k)
    # xd  = desired position
    # xdd = desired velocity
    # _   = desired acceleration (we don't use it here, _ is a "throwaway")

    # ---- Loop over all 4 followers ----
    for i in range(N_AGENTS):

        # Read this follower's current state
        x_i  = X[i, k, 0]   # current temperature (position)
        vx_i = X[i, k, 1]   # current temperature rate (velocity)

        # ------------------------------------------------------------------
        # (a)  COMPUTE TRACKING ERRORS
        # ------------------------------------------------------------------
        e1 = xd  - x_i    # position error: how far are we from desired temp?
        e2 = xdd - vx_i   # velocity error: how far is our rate from desired?

        # Store for plotting later
        E1[i, k] = e1
        E2[i, k] = e2

        # ------------------------------------------------------------------
        # (b)  DNN INFERENCE (predict f_hat)
        # ------------------------------------------------------------------
        # torch.no_grad() tells PyTorch not to track gradients here.
        # Gradients are only needed during training, not during prediction.
        # This makes inference faster and uses less memory.
        with torch.no_grad():
            # Convert Python floats to a PyTorch tensor.
            # [[x_i, vx_i]] creates a 2D tensor of shape [1, 2]
            # (1 sample, 2 features)
            inp_t = torch.tensor([[x_i, vx_i]], dtype=torch.float32)

            # Pass the input through the DNN to get f_hat
            f_hat = dnns[i](inp_t).item()
            # .item() converts the PyTorch tensor result back to a plain float

        # Store for plotting
        F_HAT[i, k]  = f_hat
        F_TRUE[i, k] = plant_f(x_i, vx_i, i)   # true value (for comparison)

        # ------------------------------------------------------------------
        # (c)  COMPUTE RISE + DNN CONTROL INPUT
        # ------------------------------------------------------------------
        u, _ = controllers[i].step(e1, e2, f_hat, DT)
        # u   = the control signal to send to the plant (in Watts)
        # _   = auxiliary error r (not used here)
        U[i, k] = u   # store for plotting

        # ------------------------------------------------------------------
        # (d)  ADVANCE THE PLANT (Euler integration)
        # ------------------------------------------------------------------
        # We simulate the cryocooler's physical response to the control input.
        #
        # Euler method: the simplest numerical integration technique.
        #   new_value = old_value + derivative * dt
        #
        # From the plant equation:  x'' = f(x, x') + u + d(t)
        #   x''  = acceleration of temperature
        #   f    = true (unknown) nonlinear term
        #   u    = our control input
        #   d    = disturbance

        f_true = plant_f(x_i, vx_i, i)   # true nonlinear term
        d_i    = disturbance(t_k, i)      # external disturbance at this moment
        accel  = f_true + u + d_i         # total acceleration (x'')

        # Update position: x[k+1] = x[k] + x'[k] * dt
        X[i, k + 1, 0] = x_i  + vx_i * DT

        # Update velocity: x'[k+1] = x'[k] + x''[k] * dt
        X[i, k + 1, 1] = vx_i + accel * DT

        # ------------------------------------------------------------------
        # (e)  ONLINE DNN TRAINING
        # ------------------------------------------------------------------
        # After every DNN_UPDATE_EVERY steps (and past the warmup period),
        # we update the DNN weights using gradient descent.
        #
        # WHAT WE KNOW:
        #   accel = f_true + u + d
        # So:
        #   accel - u = f_true + d
        #
        # We use (accel - u) as the "target" for the DNN.
        # The DNN tries to predict this from [x, x'].
        # Over time, it learns to approximate f_true (the disturbance d is
        # small and the RISE controller handles whatever the DNN misses).

        if k % DNN_UPDATE_EVERY == 0 and k >= DNN_WARMUP_STEPS:
            # k % DNN_UPDATE_EVERY == 0  means "is k divisible by 5?"
            # % is the modulo operator (remainder after division)

            # Forward pass: compute DNN prediction WITH gradient tracking
            inp_tr  = torch.tensor([[x_i, vx_i]], dtype=torch.float32)
            f_hat_t = dnns[i](inp_tr)   # shape: [1]

            # Create the supervision target: what the DNN SHOULD have output
            target = torch.tensor([[accel - u]], dtype=torch.float32)  # shape [1,1]

            # Compute loss (Mean Squared Error between prediction and target)
            # MSE = average of (prediction - target)^2
            loss = nn.functional.mse_loss(f_hat_t.unsqueeze(-1), target)
            # .unsqueeze(-1) adds a dimension: [1] -> [1,1]  (to match target shape)

            # Backward pass: compute gradients (how to change weights to reduce loss)
            optims[i].zero_grad()   # clear gradients from the previous step
            loss.backward()         # compute new gradients via backpropagation

            # Gradient clipping: prevents weights from changing too much at once.
            # If gradients are too large, clip them to magnitude 1.0
            nn.utils.clip_grad_norm_(dnns[i].parameters(), 1.0)

            # Weight update: Adam adjusts weights to reduce the loss
            optims[i].step()


# After the loop: fill in the very last time step's errors
xd_f, xdd_f, _ = leader_traj(t_vec[-1])   # t_vec[-1] = last element
for i in range(N_AGENTS):
    E1[i, -1]     = xd_f  - X[i, -1, 0]    # -1 index = last element
    E2[i, -1]     = xdd_f - X[i, -1, 1]
    F_TRUE[i, -1] = plant_f(X[i, -1, 0], X[i, -1, 1], i)

print(" done.\n")   # complete the "Running simulation ..." line


# =============================================================================
# STEP 12 — PRINT SUMMARY STATISTICS
# =============================================================================
# RMS = Root Mean Square — a standard way to measure average error size.
#   RMS = sqrt( mean( e1^2 ) )
# We skip the first 20% of the simulation (the transient start-up phase)
# so the number reflects steady-state performance only.

xd_arr  = np.array([leader_traj(t)[0] for t in t_vec])   # all desired positions
xdd_arr = np.array([leader_traj(t)[1] for t in t_vec])   # all desired velocities

rms_e1 = np.sqrt(np.mean(E1[:, N // 5:] ** 2, axis=1))
# N // 5  = skip first 20%  (integer division)
# E1[:, N//5:]  = all agents, from step N//5 to the end
# ** 2   = square each element
# np.mean(..., axis=1)  = average along time axis (axis 1)
# np.sqrt(...)  = square root

print("  Steady-state RMS position error (skip first 20%):")
for i in range(N_AGENTS):
    print(f"    Follower {i+1}: {rms_e1[i]:.4f} K")
    # f"..." = f-string: embed variables directly in text
    # :.4f  = format as a float with 4 decimal places
print()


# =============================================================================
# STEP 13 — DEFINE COLOURS AND LABELS FOR PLOTS
# =============================================================================
# We assign a distinct colour to each follower so plots are easy to read.

COLORS = plt.cm.Set1(np.linspace(0.05, 0.75, N_AGENTS))
# plt.cm.Set1   = a qualitative colour map (good for distinct categories)
# np.linspace(0.05, 0.75, 4)  = 4 evenly spaced values from 0.05 to 0.75
# These values index into the colour map to pick 4 distinct colours.

LABELS = [f"Follower {i + 1}" for i in range(N_AGENTS)]
# Creates: ["Follower 1", "Follower 2", "Follower 3", "Follower 4"]


# =============================================================================
# STEP 14 — FIGURE 1: Trajectory Tracking, Errors, and Control Inputs (2D)
# =============================================================================
# This figure has 3 sub-plots stacked vertically (3 rows, 1 column).

fig1, axs = plt.subplots(3, 1, figsize=(13, 10), sharex=True)
# plt.subplots(rows, cols)  creates a grid of subplots
# figsize=(width, height)   in inches
# sharex=True               all subplots share the same x-axis (time)
# axs is a list of 3 Axes objects: axs[0], axs[1], axs[2]

fig1.suptitle(
    "Multiagent Cryocooler - RISE + DNN  |  Leader -> 4 Followers",
    fontsize=14, fontweight="bold",
)
# suptitle = "super title" = big title above all subplots

# ---- Sub-plot 0: Position (temperature) trajectories ----
axs[0].plot(t_vec, xd_arr, "k--", lw=2.5, label="Leader (desired)", zorder=6)
# "k--"  : k=black, --=dashed line
# lw     : line width
# zorder : drawing order (higher = drawn on top)

for i in range(N_AGENTS):
    axs[0].plot(t_vec, X[i, :, 0], color=COLORS[i], alpha=0.85, label=LABELS[i])
    # X[i, :, 0]  = all time steps for follower i, position state
    # alpha=0.85  = 85% opacity (slightly transparent)

axs[0].set_ylabel("Temperature x (K)", fontsize=11)
axs[0].legend(loc="upper right", ncol=3, fontsize=9)
# legend shows coloured labels; ncol=3 arranges in 3 columns
axs[0].grid(alpha=0.3)   # faint grid lines
axs[0].set_title("State Trajectories", fontsize=11)

# ---- Sub-plot 1: Position tracking error ----
for i in range(N_AGENTS):
    axs[1].plot(t_vec, E1[i], color=COLORS[i], alpha=0.85, label=LABELS[i])

axs[1].axhline(0.0, color="k", lw=0.8, ls="--")
# axhline draws a horizontal line at y=0 (the "perfect tracking" reference)

axs[1].set_ylabel("Position error e1 (K)", fontsize=11)
axs[1].legend(loc="upper right", ncol=2, fontsize=9)
axs[1].grid(alpha=0.3)
axs[1].set_title("Position Tracking Error  e1 = xd - x", fontsize=11)

# ---- Sub-plot 2: Control input ----
for i in range(N_AGENTS):
    axs[2].plot(t_vec, U[i], color=COLORS[i], alpha=0.85, label=LABELS[i])

axs[2].set_xlabel("Time (s)", fontsize=11)
axs[2].set_ylabel("Control input u (W)", fontsize=11)
axs[2].legend(loc="upper right", ncol=2, fontsize=9)
axs[2].grid(alpha=0.3)
axs[2].set_title("RISE + DNN Control Inputs", fontsize=11)

plt.tight_layout()   # automatically adjust spacing so labels don't overlap
fig1.savefig("fig1_tracking.png", dpi=150, bbox_inches="tight")
# dpi=150           = dots per inch (higher = sharper image)
# bbox_inches="tight" = crop whitespace around the figure


# =============================================================================
# STEP 15 — FIGURE 2: DNN Approximation Quality (2D)
# =============================================================================
# For each follower, compare what the DNN predicts (f_hat) vs the true f(x,x').
# This shows whether the neural network has learned the plant dynamics.

fig2, axs2 = plt.subplots(2, 2, figsize=(13, 8))
# 2 rows x 2 columns = 4 subplots (one per follower)
# axs2 is a 2D array: axs2[row, col]

fig2.suptitle(
    "DNN Online Approximation of Plant Nonlinearity  f(x, x')",
    fontsize=13, fontweight="bold",
)

for i in range(N_AGENTS):
    ax = axs2[i // 2, i % 2]
    # i=0 -> row 0, col 0    (top-left)
    # i=1 -> row 0, col 1    (top-right)
    # i=2 -> row 1, col 0    (bottom-left)
    # i=3 -> row 1, col 1    (bottom-right)

    ax.plot(t_vec, F_TRUE[i], "k-", lw=1.5, label="True  f(x,x')", alpha=0.85)
    ax.plot(t_vec, F_HAT[i],  "--", lw=1.5, label="DNN   f_hat(x,x')",
            color=COLORS[i], alpha=0.90)

    ax.set_title(
        f"Follower {i+1}  "
        f"(d={delta_arr[i]:.3f}, a={alpha_arr[i]:.3f}, b={beta_arr[i]:.3f})",
        fontsize=10
    )
    ax.set_xlabel("Time (s)")
    ax.set_ylabel("f(x, x')")
    ax.legend(fontsize=9)
    ax.grid(alpha=0.3)

plt.tight_layout()
fig2.savefig("fig2_dnn_approx.png", dpi=150, bbox_inches="tight")


# =============================================================================
# STEP 16 — FIGURE 3: 3D TRACKING ERROR VISUALISATION (main result)
# =============================================================================
# This is the key figure with four 3D sub-plots showing the tracking error
# across both time and agent dimensions simultaneously.
#
# Downsampling: 3D surface plots with 5000 points per agent are slow.
# We take every 8th point (DS=8) to reduce it to ~625 points per agent,
# which renders much faster with negligible loss of visual detail.

DS    = 8              # downsample factor (take every 8th time step)
t_ds  = t_vec[::DS]   # downsampled time vector
E1_ds = E1[:, ::DS]   # downsampled position errors  shape: (4, ~625)
E2_ds = E2[:, ::DS]   # downsampled velocity errors
En_ds = np.sqrt(E1_ds ** 2 + E2_ds ** 2)   # error norm at each point

# Create a grid (mesh) needed for 3D surface plots.
# agent_ax = [1, 2, 3, 4]  (follower index on the y-axis)
agent_ax = np.arange(1, N_AGENTS + 1)

# np.meshgrid creates two 2D arrays from two 1D arrays.
# T_grid[i,j] = t_ds[j]     (time value)
# A_grid[i,j] = agent_ax[i] (agent index)
# Together they define every (time, agent) coordinate on the 3D surface.
T_grid, A_grid = np.meshgrid(t_ds, agent_ax)

# Create the figure (large, to fit 4 subplots)
fig3 = plt.figure(figsize=(17, 13))
fig3.suptitle(
    "3-D Tracking Error  -  Multiagent Cryocooler  (RISE + DNN)",
    fontsize=15, fontweight="bold", y=0.99,
)

# ---- Panel A: Position Error Surface ----------------------------------------
# projection="3d" activates the 3D axes (registered by importing Axes3D above)
ax_A = fig3.add_subplot(221, projection="3d")
# 221 = 2 rows, 2 cols, subplot 1 (top-left)

surf_A = ax_A.plot_surface(
    T_grid, A_grid, E1_ds,
    cmap="coolwarm",     # blue=negative error, red=positive error
    alpha=0.88,          # slight transparency to see the shape better
    linewidth=0,         # no grid lines on the surface
    antialiased=True,    # smooth edges
)
ax_A.set_xlabel("Time (s)",       fontsize=9, labelpad=7)
ax_A.set_ylabel("Follower Agent", fontsize=9, labelpad=7)
ax_A.set_zlabel("e1  (K)",        fontsize=9, labelpad=7)
ax_A.set_yticks(agent_ax)
ax_A.set_yticklabels([f"F{j}" for j in agent_ax])  # label as F1, F2, F3, F4
ax_A.set_title("(A)  Position Error  e1", fontsize=11, pad=10)
fig3.colorbar(surf_A, ax=ax_A, shrink=0.55, pad=0.12, label="e1  (K)")
# colorbar adds the colour legend bar beside the 3D surface

# ---- Panel B: Velocity Error Surface ----------------------------------------
ax_B = fig3.add_subplot(222, projection="3d")   # 222 = top-right

surf_B = ax_B.plot_surface(
    T_grid, A_grid, E2_ds,
    cmap="RdYlGn",   # Red=large error, Yellow=medium, Green=small
    alpha=0.88, linewidth=0, antialiased=True,
)
ax_B.set_xlabel("Time (s)",       fontsize=9, labelpad=7)
ax_B.set_ylabel("Follower Agent", fontsize=9, labelpad=7)
ax_B.set_zlabel("e2  (K/s)",      fontsize=9, labelpad=7)
ax_B.set_yticks(agent_ax)
ax_B.set_yticklabels([f"F{j}" for j in agent_ax])
ax_B.set_title("(B)  Velocity Error  e2", fontsize=11, pad=10)
fig3.colorbar(surf_B, ax=ax_B, shrink=0.55, pad=0.12, label="e2  (K/s)")

# ---- Panel C: 3D Phase Portrait (time x e1 x e2) ----------------------------
# A phase portrait shows how two error signals (e1 and e2) evolve together.
# If the controller works, both should converge to zero — the lines spiral
# toward the origin (0, 0) over time.

ax_C = fig3.add_subplot(223, projection="3d")   # 223 = bottom-left

for i in range(N_AGENTS):
    # Draw the trajectory of each follower as a 3D line
    ax_C.plot(
        t_ds, E1_ds[i], E2_ds[i],
        color=COLORS[i], lw=1.8, alpha=0.85, label=LABELS[i],
    )
    # Mark the starting point (large circle)
    ax_C.scatter(t_ds[0], E1_ds[i, 0], E2_ds[i, 0],
                 color=COLORS[i], s=70, zorder=6)
    # Mark the ending point (star marker)
    ax_C.scatter(t_ds[-1], E1_ds[i, -1], E2_ds[i, -1],
                 color=COLORS[i], s=60, marker="*", zorder=6)

ax_C.set_xlabel("Time (s)",  fontsize=9, labelpad=7)
ax_C.set_ylabel("e1  (K)",   fontsize=9, labelpad=7)
ax_C.set_zlabel("e2  (K/s)", fontsize=9, labelpad=7)
ax_C.set_title("(C)  Error Phase Portrait  (all followers)", fontsize=11, pad=10)
ax_C.legend(loc="upper left", fontsize=8, framealpha=0.75)

# ---- Panel D: Error Norm Surface --------------------------------------------
# The error norm ||[e1, e2]||2 = sqrt(e1^2 + e2^2) combines both errors
# into one number measuring overall tracking quality.
# A flat surface near zero = excellent tracking.

ax_D = fig3.add_subplot(224, projection="3d")   # 224 = bottom-right

surf_D = ax_D.plot_surface(
    T_grid, A_grid, En_ds,
    cmap="viridis",   # dark purple=small, yellow=large
    alpha=0.88, linewidth=0, antialiased=True,
)
ax_D.set_xlabel("Time (s)",       fontsize=9, labelpad=7)
ax_D.set_ylabel("Follower Agent", fontsize=9, labelpad=7)
ax_D.set_zlabel("||e||2  (K)",    fontsize=9, labelpad=7)
ax_D.set_yticks(agent_ax)
ax_D.set_yticklabels([f"F{j}" for j in agent_ax])
ax_D.set_title("(D)  Error Norm  ||[e1, e2]||2", fontsize=11, pad=10)
fig3.colorbar(surf_D, ax=ax_D, shrink=0.55, pad=0.12, label="||e||2  (K)")

plt.tight_layout(rect=[0.0, 0.0, 1.0, 0.97])
# rect=[left, bottom, right, top] reserves space for the suptitle
fig3.savefig("fig3_3d_error.png", dpi=150, bbox_inches="tight")


# =============================================================================
# STEP 17 — FIGURE 4: Error Convergence (Log Scale + Linear Scale)
# =============================================================================
# Two sub-plots showing how the error norm decreases over time.
#
# Log scale is useful because it reveals whether the error decays
# exponentially (a straight line on a log plot = exponential decay).

E_NORM_2D = np.sqrt(E1 ** 2 + E2 ** 2)   # error norm at all time steps

fig4, axs4 = plt.subplots(2, 1, figsize=(13, 7), sharex=True)
fig4.suptitle("Error Norm Convergence - All Followers",
              fontsize=13, fontweight="bold")

# ---- Sub-plot 0: Log scale ----
axs4[0].set_yscale("log")   # change y-axis to logarithmic scale
for i in range(N_AGENTS):
    axs4[0].plot(t_vec, E_NORM_2D[i], color=COLORS[i], alpha=0.85,
                 label=LABELS[i])
axs4[0].set_ylabel("||[e1, e2]||2  (log scale)", fontsize=11)
axs4[0].legend(ncol=2, fontsize=9)
axs4[0].grid(alpha=0.3, which="both")   # which="both" adds grid on minor ticks too
axs4[0].set_title("Tracking Error Norm (log scale shows convergence rate)",
                  fontsize=10)

# ---- Sub-plot 1: Linear scale ----
for i in range(N_AGENTS):
    axs4[1].plot(t_vec, E1[i], color=COLORS[i], alpha=0.85, label=LABELS[i])
axs4[1].axhline(0.0, color="k", lw=0.8, ls="--")
axs4[1].set_xlabel("Time (s)", fontsize=11)
axs4[1].set_ylabel("Position error e1 (K)", fontsize=11)
axs4[1].legend(ncol=2, fontsize=9)
axs4[1].grid(alpha=0.3)
axs4[1].set_title("Position Error  e1  (linear scale)", fontsize=10)

plt.tight_layout()
fig4.savefig("fig4_convergence.png", dpi=150, bbox_inches="tight")


# =============================================================================
# STEP 18 — SHOW ALL FIGURES
# =============================================================================
print("Saved:  fig1_tracking.png")
print("        fig2_dnn_approx.png")
print("        fig3_3d_error.png  <-- main 3-D visualisation")
print("        fig4_convergence.png")
print("\nShowing figures ...")

plt.show()
# plt.show() displays all figures that were created.
# The program pauses here until you close the figure windows.
# If running in a terminal with no display (e.g. a server),
# set the environment variable: MPLBACKEND=Agg
# and figures will be saved as PNG files without displaying a window.
```
